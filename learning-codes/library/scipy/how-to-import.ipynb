## 1. scipy.optimize – Optimization
- Used to find the minimum or maximum of a function.
- Helps solve mathematical optimization problems.
- Example: Finding the minimum cost or error in a machine learning model.
## 2. scipy.stats – Statistics
- Used for statistical calculations.
- Provides probability distributions and statistical tests.
- Example: Calculating probability, mean, standard deviation and hypothesis testing.
## 3. scipy.linalg – Linear Algebra
- Used to perform advanced linear algebra operations.
- Works with matrices and vectors.
- Example: Matrix inverse, determinant, eigenvalues and solving linear equations.
## 4. scipy.optimize.minimize – Minimize Function
- Imports only the minimize() function.
- Used to find the minimum value of a mathematical function.

```python
# Importing the minimize function
from scipy.optimize import minimize

# Defining a mathematical function
def objective(x):
    return x**2 + 4*x + 4

# Finding the minimum value
result = minimize(objective, x0=0)

# Displaying the result
print(result.x)
print(result.fun)
```

```python
from scipy import optimize

# find the minimum of f(x) = (x - 3)^2 + 1
result = optimize.minimize_scalar(lambda x: (x - 3)**2 + 1)
print(result.x)     # about 3.0
print(result.fun)   # about 1.0
```
