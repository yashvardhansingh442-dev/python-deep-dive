```python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5, 6]

# Positive relationship
positive_y = [10, 20, 30, 40, 50, 60]

# Negative relationship
negative_y = [60, 50, 40, 30, 20, 10]

# No clear linear relationship
random_y = [25, 55, 15, 60, 35, 45]

# Creating three scatter plots
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# Positive correlation
axes[0].scatter(x, positive_y)
axes[0].set_title("Positive Correlation")

# Negative correlation
axes[1].scatter(x, negative_y)
axes[1].set_title("Negative Correlation")

# No clear linear correlation
axes[2].scatter(x, random_y)
axes[2].set_title("No Clear Linear Trend")

# Displaying all three plots
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Creating the data
activities = [
    "Sleep",
    "College",
    "Study",
    "Exercise",
    "Entertainment",
    "Other"
]

hours = [8, 6, 4, 1, 3, 2]

# Creating the chart
plt.figure(figsize=(8, 8))

# Creating the donut chart
plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

# Adding title
plt.title("My Daily Time Distribution")

# Keeping the chart circular
plt.axis("equal")

# Saving the chart
plt.savefig("daily_time.png", dpi=300, bbox_inches="tight")

# Displaying the chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]
y3 = [10, 25, 15, 35, 30]
y4 = [5, 15, 25, 35, 45]

# Creating a 2x2 subplot layout
fig, axes = plt.subplots(2, 2)

# First graph
axes[0, 0].plot(x, y1)
axes[0, 0].set_title("Increasing Data")

# Second graph
axes[0, 1].plot(x, y2)
axes[0, 1].set_title("Decreasing Data")

# Third graph
axes[1, 0].plot(x, y3)
axes[1, 0].set_title("Variable Data")

# Fourth graph
axes[1, 1].plot(x, y4)
axes[1, 1].set_title("Another Dataset")

# Adjusting the spacing
plt.tight_layout()

# Displaying the figure
plt.show()
```

```python
import matplotlib.pyplot as plt

# Creating the data
subjects = ["Math", "Physics", "Python", "Statistics"]

marks = [85, 75, 90, 80]
hours = [5, 3, 6, 4]

# Creating two subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Creating a bar chart
axes[0].bar(subjects, marks, color="skyblue")
axes[0].set_title("Subject Marks")
axes[0].set_ylabel("Marks")

# Creating a line chart
axes[1].plot(subjects, hours, marker="o", color="orange")
axes[1].set_title("Study Hours")
axes[1].set_ylabel("Hours")

# Rotating category labels
axes[0].tick_params(axis="x", rotation=30)
axes[1].tick_params(axis="x", rotation=30)

# Adjusting spacing
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [15, 25, 35, 45, 55]

# Creating two subplots with shared axes
fig, axes = plt.subplots(
    2,
    1,
    sharex=True,
    sharey=True
)

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Dataset A")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Dataset B")

# Adding a shared X-axis label
fig.supxlabel("X Values")

# Adding a shared Y-axis label
fig.supylabel("Y Values")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]

# Creating two plots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Increasing")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Decreasing")

# Adding the main figure title
fig.suptitle("Data Comparison")

# Adjusting layout to make room for the main title
fig.tight_layout(rect=[0, 0, 1, 0.93])

plt.show()
```
