```python
import numpy as np
from scipy.sparse import csr_array

# Customer IDs
customers = np.array([0, 1, 2, 2, 3, 4, 4])

# Product IDs
products = np.array([1, 1, 2, 3, 1, 4, 5])

# Interaction values
interactions = np.ones(len(customers))

# Creating a sparse matrix
X = csr_array(
    (interactions, (customers, products)),
    shape=(5, 6)
)
```
