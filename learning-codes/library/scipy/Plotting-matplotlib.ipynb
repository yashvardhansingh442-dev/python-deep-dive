## 1. Setup & a quick taste

**Matplotlib** is Python's foundational plotting library. Almost every other plotting tool (Seaborn, pandas `.plot()`) is built on top of it.

By convention we import two things:

| Import | Alias | Purpose |
|---|---|---|
| `matplotlib.pyplot` | `plt` | The plotting interface |
| `numpy` | `np` | Fast numeric arrays, handy for generating data |

Install if needed: `pip install matplotlib numpy`

Below is a preview of what we'll be able to do — two curves on one set of axes with labels, a legend, and a grid. Don't worry about every line yet; each piece is explained in the sections that follow.

> 💡 `np.linspace(0, 2π, 200)` creates 200 evenly spaced points between 0 and 2π, which makes the curves look smooth.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
````

**What happened here**
- `plt.plot(x, y, label=...)` draws a line. Calling it twice puts two lines on the same axes.
- `plt.legend()` uses each line's `label` to build the legend.
- `plt.grid(True)` adds background grid lines.
- `plt.show()` renders the figure.

## 2. Core concepts

### The anatomy of a plot
- **Figure** — the whole canvas (the "window" or image).
- **Axes** — a single plot area *inside* the figure, with its own x-axis, y-axis, title, etc. (A figure can contain many axes.)
- **Axis** — the x or y number line itself (ticks and labels).

### Two ways to use Matplotlib
| Style | Looks like | Best for |
|---|---|---|
| **pyplot (state-based)** | `plt.plot(x, y)` | Quick, simple plots |
| **Object-oriented (OO)** | `fig, ax = plt.subplots()` then `ax.plot(x, y)` | Multiple plots, custom layouts, real projects |

Tip: start with pyplot, but get comfortable with the OO style early — it scales much better.

### 2.1 Your first line plot (pyplot style)

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating a line plot
plt.plot(x, y)

# Displaying the graph
plt.show()
````

Here `x` and `y` are plain Python lists. Matplotlib pairs them up as points `(1,2), (2,4), ...` and connects them with a line.

### 2.2 The same idea, object-oriented style, with a custom size

`plt.subplots()` returns **two things**: the `fig` (canvas) and `ax` (the plot area). `figsize=(width, height)` is measured in inches.

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating figure with custom dimensions
fig, ax = plt.subplots(figsize=(10, 5))

# Creating data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating line plot
ax.plot(x, y)

# Displaying graph
plt.show()
````

> 🔑 **Rule of thumb:** with the OO style, methods that were `plt.title()` / `plt.xlabel()` become `ax.set_title()` / `ax.set_xlabel()`.

**Try it yourself:** change `figsize` to `(4, 4)` and then `(12, 3)`. How does the shape of the line change?

## 3. Line plots

Line plots are best for showing **trends over time** or any continuous relationship.

### 3.1 Styling a line: `color`, `linestyle`, `marker`

| Parameter | Examples |
|---|---|
| `color` | `"red"`, `"blue"`, `"#ff8800"` |
| `linestyle` | `"-"` solid, `"--"` dashed, `":"` dotted, `"-."` dash-dot |
| `marker` | `"o"` circle, `"s"` square, `"^"` triangle, `"*"` star |

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating a red dashed line with circle markers
plt.plot(x, y, color="red", linestyle="--", marker="o")

# Displaying the graph
plt.show()
````

### 3.2 Adding a title and axis labels

A graph without labels is hard to read. Always add:
- `plt.title()` — what the chart is about
- `plt.xlabel()` / `plt.ylabel()` — what each axis measures (include units!)

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]
sales = [100, 150, 130, 200, 250, 300]

# Creating the line plot
plt.plot(months, sales, color="blue", marker="o")

# Adding graph title
plt.title("Monthly Sales")

# Naming the X-axis
plt.xlabel("Months")

# Naming the Y-axis
plt.ylabel("Sales")

# Displaying the graph
plt.show()
````

### 3.3 Plotting multiple lines with a legend

Give each line a `label`, then call `plt.legend()`. Using different colours **and** different markers (`"o"` vs `"s"`) keeps the chart readable even when printed in black and white.

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]

product_a = [100, 150, 130, 200, 250, 300]
product_b = [80, 120, 160, 170, 220, 280]

# Plotting Product A
plt.plot(months, product_a, color="blue", marker="o", label="Product A")

# Plotting Product B
plt.plot(months, product_b, color="red", marker="s", label="Product B")

# Adding title and axis labels
plt.title("Product Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

# Displaying the legend
plt.legend()

# Displaying the graph
plt.show()
````

### 3.4 Full styling: line width, marker size, grid

Extra tweaks:
- `linewidth` — thickness of the line
- `markersize` — size of the markers
- `plt.grid(True, alpha=0.4)` — `alpha` controls transparency (0 = invisible, 1 = solid), so the grid stays subtle

````python
import matplotlib.pyplot as plt

# Creating the data
days = [1, 2, 3, 4, 5, 6, 7]

temperature = [25, 27, 26, 29, 31, 30, 32]

# Creating the line plot
plt.plot(
    days,
    temperature,
    color="orange",
    linestyle="--",
    marker="o",
    linewidth=2,
    markersize=7,
    label="Temperature"
)

# Adding title and axis labels
plt.title("Weekly Temperature Report")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

# Adding legend and grid
plt.legend()
plt.grid(True, alpha=0.4)

# Displaying the graph
plt.show()
````

**Try it yourself:** plot your own 7-day data (e.g. steps walked or hours slept). Use a green dotted line with star markers.

## 4. Bar charts

Bar charts **compare values across categories** (e.g. sales per product).

### 4.1 Basic vertical bar chart
`plt.bar(categories, values)`

````python
import matplotlib.pyplot as plt

# Creating the data
products = ["Laptop", "Mobile", "Tablet", "Watch", "Camera"]
sales = [120, 200, 90, 150, 70]

# Creating the bar chart
plt.bar(products, sales)

# Adding title and labels
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

# Displaying the graph
plt.show()
````

### 4.2 Different colour per bar
Pass a **list** of colours — one for each bar, in order.

````python
import matplotlib.pyplot as plt

# Creating the data
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 200, 90, 150]

# Creating a bar chart with different colors
plt.bar(
    products,
    sales,
    color=["red", "blue", "green", "orange"]
)

# Adding title
plt.title("Product Sales")

# Displaying the graph
plt.show()
````

### 4.3 Horizontal bars with `barh`
Use `plt.barh()` when category names are long, or when you want to rank items. Note that the roles of the axes are swapped: categories go on the **y-axis**, values on the **x-axis**.

````python
import matplotlib.pyplot as plt

# Creating the data
subjects = ["Math", "Physics", "Python", "Statistics"]
hours = [5, 3, 6, 4]

# Creating a horizontal bar chart
plt.barh(subjects, hours, color="purple")

# Adding title and labels
plt.title("Study Hours")
plt.xlabel("Hours")
plt.ylabel("Subjects")

# Displaying the graph
plt.show()
````

### 4.4 Showing the value on each bar
`ax.bar_label(bars)` writes the number on top of every bar so readers don't have to estimate from the axis. This uses the OO style: `ax.bar()` *returns* the bars, which we hand to `bar_label`.

````python
import matplotlib.pyplot as plt

# Creating the data
products = ["Laptop", "Mobile", "Tablet"]
sales = [120, 200, 90]

# Creating the bar chart
fig, ax = plt.subplots()

bars = ax.bar(products, sales, color="skyblue")

# Displaying values above the bars
ax.bar_label(bars)

# Adding title
ax.set_title("Product Sales")

plt.show()
````

## 5. Histograms

A histogram shows the **distribution** of one numeric variable: values are grouped into **bins**, and the bar height is how many values fall in each bin.

> 🆚 **Bar chart vs histogram:** a bar chart compares *categories*; a histogram groups a *continuous number* into ranges. The bars in a histogram touch each other.

### 5.1 Histogram of student marks
- `bins=5` → split the data range into 5 equal groups
- `edgecolor="black"` → draws outlines so individual bars are easy to see

````python
import matplotlib.pyplot as plt

# Creating sample marks of students
marks = [
    35, 40, 45, 50, 55, 60, 65, 70,
    72, 75, 78, 80, 82, 85, 88, 90,
    92, 95, 45, 55, 60, 65, 70, 75
]

# Creating the histogram
plt.hist(marks, bins=5, color="skyblue", edgecolor="black")

# Adding title and labels
plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

# Displaying the graph
plt.show()
````

### 5.2 More data, more bins
Here we generate 1000 random marks from a **normal distribution** (mean 70, standard deviation 10) using `np.random.normal(70, 10, 1000)`. With plenty of data, more bins reveal the classic **bell curve** shape.

> ⚠️ The data is random, so your plot will look slightly different each run. Add `np.random.seed(42)` before generating data if you want identical results every time.

> 🎯 **Choosing bins:** too few bins hide the shape; too many make it noisy. Try 5, 20 and 50 on the same data and compare.

````python
import numpy as np
import matplotlib.pyplot as plt

# Generating 1000 random marks
marks = np.random.normal(70, 10, 1000)

# Creating the histogram
plt.hist(
    marks,
    bins=20,
    color="orange",
    edgecolor="black"
)

# Adding title and labels
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Displaying the graph
plt.show()
````

**Try it yourself:** change `bins` to 5, then 50. Which one tells the clearest story?

## 6. Scatter plots

A scatter plot shows each observation as a **dot** at position `(x, y)`. It is the go-to chart for exploring the **relationship between two variables**.

### 6.1 Study hours vs exam marks

````python
import matplotlib.pyplot as plt

# Creating the data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
exam_marks = [35, 40, 45, 55, 60, 70, 75, 90]

# Creating the scatter plot
plt.scatter(study_hours, exam_marks)

# Adding title and labels
plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

# Displaying the graph
plt.show()
````

The dots rise from left to right: more study hours go with higher marks — a **positive relationship**.

### 6.2 Styling a scatter plot
Same idea, with `color="purple"`. (You can also control dot size with `s=` and transparency with `alpha=`.)

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5, 6]
y = [10, 20, 25, 40, 45, 60]

# Creating a scatter plot
plt.scatter(x, y, color="purple")

plt.show()
````

### 6.3 Reading correlation: positive, negative, none

| Pattern | Meaning | Example |
|---|---|---|
| Dots go **up** → | Positive correlation | Study hours ↑, marks ↑ |
| Dots go **down** → | Negative correlation | Phone time ↑, sleep ↓ |
| **No clear shape** | No linear relationship | Shoe size vs exam marks |

> ⚠️ **Correlation ≠ causation.** Two things moving together doesn't prove one causes the other.

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5, 6]

# Positive relationship
positive_y = [10, 20, 30, 40, 50, 60]

# Negative relationship
negative_y = [60, 50, 40, 30, 20, 10]

# No clear linear relationship
random_y = [25, 55, 15, 60, 35, 45]

# Creating three scatter plots
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# Positive correlation
axes[0].scatter(x, positive_y)
axes[0].set_title("Positive Correlation")

# Negative correlation
axes[1].scatter(x, negative_y)
axes[1].set_title("Negative Correlation")

# No clear linear correlation
axes[2].scatter(x, random_y)
axes[2].set_title("No Clear Linear Trend")

# Displaying all three plots
plt.tight_layout()
plt.show()
````

**Try it yourself:** scatter `size=[...]` vs `price=[...]` for 8 houses, with `s=100` and `alpha=0.6`.

## 7. Pie & donut charts

Pie charts show **parts of a whole** (shares add up to 100%). They work best with **few categories** (about 5 or fewer).

### 7.1 Basic pie chart

````python
import matplotlib.pyplot as plt

# Creating the data
activities = ["Sleep", "Study", "Work", "Exercise", "Free Time"]
hours = [8, 5, 6, 2, 3]

# Creating the pie chart
plt.pie(hours, labels=activities)

# Adding title
plt.title("Daily Time Distribution")

# Displaying the chart
plt.show()
````

### 7.2 Showing percentages with `autopct`
`autopct="%1.1f%%"` is a format string: show each slice's percentage with 1 decimal place. (`%%` prints a literal `%` sign.)

````python
import matplotlib.pyplot as plt

# Creating the data
activities = ["Sleep", "Study", "Work", "Exercise", "Free Time"]
hours = [8, 5, 6, 2, 3]

# Creating the pie chart
plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%"
)

# Adding title
plt.title("Daily Time Distribution")

plt.show()
````

### 7.3 Custom colours

````python
import matplotlib.pyplot as plt

# Creating the data
languages = ["Python", "Java", "C++", "JavaScript"]
popularity = [40, 25, 20, 15]

# Defining colors
colors = ["skyblue", "orange", "lightgreen", "pink"]

# Creating the pie chart
plt.pie(
    popularity,
    labels=languages,
    colors=colors,
    autopct="%1.1f%%"
)

plt.title("Programming Language Popularity")
plt.show()
````

### 7.4 Highlighting a slice with `explode`
`explode` is a list — one number per slice — that says how far to pull each slice out from the centre. `0.1` pulls the first slice out a little.

````python
import matplotlib.pyplot as plt

# Creating the data
languages = ["Python", "Java", "C++", "JavaScript"]
popularity = [40, 25, 20, 15]

# Moving the first slice outward
explode = [0.1, 0, 0, 0]

# Creating the pie chart
plt.pie(
    popularity,
    labels=languages,
    explode=explode,
    autopct="%1.1f%%"
)

plt.title("Programming Language Popularity")
plt.show()
````

### 7.5 Donut chart
A donut is just a pie with a hole. Setting `wedgeprops={"width": 0.4}` limits each wedge's thickness, leaving the middle empty. Donuts are often easier to read than full pies.

````python
import matplotlib.pyplot as plt

# Creating the data
categories = ["Python", "Java", "C++", "JavaScript"]
values = [40, 25, 20, 15]

# Creating a donut chart
plt.pie(
    values,
    labels=categories,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.title("Programming Language Distribution")

plt.show()
````

### 7.6 Using a legend instead of labels on slices
When labels would overlap on the slices, move them into a legend with `plt.legend(labels, title=...)`.

````python
import matplotlib.pyplot as plt

# Creating the data
subjects = ["Math", "Physics", "Python", "Statistics"]
hours = [5, 3, 6, 4]

# Creating the pie chart
plt.pie(hours)

# Adding a legend
plt.legend(subjects, title="Subjects")

# Adding title
plt.title("Study Time Distribution")

plt.show()
````

**Try it yourself:** build a donut chart of how you spend a week's budget, with percentages and the biggest slice exploded.

## 8. Annotations

`plt.annotate()` lets you point at something important on the chart with text and an arrow.

| Argument | Meaning |
|---|---|
| first argument | The text to show |
| `xy=(x, y)` | The point being **pointed at** |
| `xytext=(x, y)` | Where the **text** is placed |
| `arrowprops` | Style of the arrow, e.g. `{"arrowstyle": "->"}` |

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]
sales = [100, 150, 130, 200, 250, 300]

# Creating the line plot
plt.plot(months, sales, marker="o")

# Adding annotation
plt.annotate(
    "Highest Sales",
    xy=(6, 300),
    xytext=(4, 320),
    arrowprops={"arrowstyle": "->"}
)

# Adding title and labels
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
````

**Try it yourself:** annotate the *lowest* sales month instead of the highest.

## 9. Saving figures

Use `plt.savefig("name.png")` to save a plot as an image for reports, slides or GitHub READMEs.

| Option | Purpose |
|---|---|
| `dpi=300` | Resolution (higher = sharper; 300 is print quality) |
| `bbox_inches="tight"` | Trims extra whitespace and stops labels getting cut off |

> ⚠️ **Call `savefig()` *before* `plt.show()`.** After `show()`, the figure is cleared and you may save a blank image.

Supported formats include `.png`, `.jpg`, `.pdf` and `.svg` (the extension decides the format).

### 9.1 Save a simple line plot

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

# Creating the plot
plt.plot(x, y, marker="o")

# Adding title
plt.title("My First Saved Graph")

# Saving the graph
plt.savefig("my_graph.png", dpi=300, bbox_inches="tight")

# Displaying the graph
plt.show()
````

### 9.2 Putting it together: a saved donut chart of a daily routine
This combines several ideas from earlier: `figsize`, `autopct`, a donut via `wedgeprops`, `startangle=90` (start the first slice at the top), and `plt.axis("equal")` which keeps the circle perfectly round.

````python
import matplotlib.pyplot as plt

# Creating the data
activities = [
    "Sleep",
    "College",
    "Study",
    "Exercise",
    "Entertainment",
    "Other"
]

hours = [8, 6, 4, 1, 3, 2]

# Creating the chart
plt.figure(figsize=(8, 8))

# Creating the donut chart
plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

# Adding title
plt.title("My Daily Time Distribution")

# Keeping the chart circular
plt.axis("equal")

# Saving the chart
plt.savefig("daily_time.png", dpi=300, bbox_inches="tight")

# Displaying the chart
plt.show()
````

## 10. Subplots

Subplots put **multiple charts in one figure**. `plt.subplots(rows, cols)` returns the figure and an array of axes.

- 1 row × 2 columns → `axes[0]`, `axes[1]`
- 2 rows × 2 columns → `axes[row, col]` e.g. `axes[0, 1]`

### 10.1 Two plots side by side

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating two plotting areas
fig, axes = plt.subplots(1, 2)

# First dataset
x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]

# Second dataset
y2 = [1, 4, 9, 16]

# Plotting on first axes
axes[0].plot(x, y1)

# Plotting on second axes
axes[1].plot(x, y2)

# Displaying figure
plt.show()
````

Here `axes[0]` is the left plot and `axes[1]` the right. The second dataset (`1, 4, 9, 16`) is the squares, so you see a curve instead of a straight line.

### 10.2 Comparing the same measure across years

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5]

sales_2025 = [100, 150, 200, 180, 250]
sales_2026 = [120, 170, 220, 240, 300]

# Creating one row and two columns of plots
fig, axes = plt.subplots(1, 2)

# Creating the first graph
axes[0].plot(months, sales_2025)
axes[0].set_title("Sales 2025")

# Creating the second graph
axes[1].plot(months, sales_2026)
axes[1].set_title("Sales 2026")

# Displaying both graphs
plt.show()
````

⚠️ **Watch out:** the two plots may have *different* y-axis ranges, which can mislead the eye. Section 10.5 shows how to fix that with `sharey`.

### 10.3 A 2×2 grid
With two indices, `axes[row, col]`. `plt.tight_layout()` adjusts spacing so titles and labels don't overlap.

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]
y3 = [10, 25, 15, 35, 30]
y4 = [5, 15, 25, 35, 45]

# Creating a 2x2 subplot layout
fig, axes = plt.subplots(2, 2)

# First graph
axes[0, 0].plot(x, y1)
axes[0, 0].set_title("Increasing Data")

# Second graph
axes[0, 1].plot(x, y2)
axes[0, 1].set_title("Decreasing Data")

# Third graph
axes[1, 0].plot(x, y3)
axes[1, 0].set_title("Variable Data")

# Fourth graph
axes[1, 1].plot(x, y4)
axes[1, 1].set_title("Another Dataset")

# Adjusting the spacing
plt.tight_layout()

# Displaying the figure
plt.show()
````

### 10.4 Mixing chart types in one figure
Each axes is independent, so you can combine a bar chart and a line chart. `tick_params(axis="x", rotation=30)` rotates the category names so they don't collide.

````python
import matplotlib.pyplot as plt

# Creating the data
subjects = ["Math", "Physics", "Python", "Statistics"]

marks = [85, 75, 90, 80]
hours = [5, 3, 6, 4]

# Creating two subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Creating a bar chart
axes[0].bar(subjects, marks, color="skyblue")
axes[0].set_title("Subject Marks")
axes[0].set_ylabel("Marks")

# Creating a line chart
axes[1].plot(subjects, hours, marker="o", color="orange")
axes[1].set_title("Study Hours")
axes[1].set_ylabel("Hours")

# Rotating category labels
axes[0].tick_params(axis="x", rotation=30)
axes[1].tick_params(axis="x", rotation=30)

# Adjusting spacing
plt.tight_layout()

plt.show()
````

### 10.5 Shared axes with `sharex` / `sharey`
When plots are meant to be **compared**, share the axes so they use the same scale. `fig.supxlabel()` and `fig.supylabel()` add one label for the whole figure instead of repeating it on every plot.

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [15, 25, 35, 45, 55]

# Creating two subplots with shared axes
fig, axes = plt.subplots(
    2,
    1,
    sharex=True,
    sharey=True
)

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Dataset A")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Dataset B")

# Adding a shared X-axis label
fig.supxlabel("X Values")

# Adding a shared Y-axis label
fig.supylabel("Y Values")

plt.tight_layout()

plt.show()
````

### 10.6 A main title for the whole figure
- `fig.suptitle()` — title for the entire figure (each axes can still have its own `set_title`)
- `rect=[0, 0, 1, 0.93]` in `tight_layout` reserves the top 7% of the figure so the main title isn't squashed

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]

# Creating two plots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Increasing")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Decreasing")

# Adding the main figure title
fig.suptitle("Data Comparison")

# Adjusting layout to make room for the main title
fig.tight_layout(rect=[0, 0, 1, 0.93])

plt.show()
````

## 11. Choosing the right chart

| Question you want to answer | Chart | Function |
|---|---|---|
| How does something **change over time**? | Line | `plt.plot` |
| How do **categories compare**? | Bar | `plt.bar` / `plt.barh` |
| How is **one variable distributed**? | Histogram | `plt.hist` |
| Is there a **relationship between two variables**? | Scatter | `plt.scatter` |
| What **share** does each part have of a whole? | Pie / Donut | `plt.pie` |
| I need **several views side by side** | Subplots | `plt.subplots` |

## 13. Common mistakes

1. **Mismatched lengths** — `x` and `y` must have the same number of items, otherwise you get a `ValueError`.
2. **`savefig` after `show`** — gives a blank image. Save first.
3. **Forgetting labels and titles** — a chart nobody can interpret is a failed chart.
4. **Too many pie slices** — switch to a bar chart if you have more than ~5 categories.
5. **Overlapping text in subplots** — add `plt.tight_layout()`.
6. **Truncated y-axis on bar charts** — bars should start at 0, or they exaggerate differences.
7. **Misleading comparisons across subplots** — use `sharey=True` when the scales should match.
8. **Mixing pyplot and OO styles carelessly** — pick one per figure.
