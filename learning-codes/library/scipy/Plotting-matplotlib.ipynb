# Matplotlib basics

Core plotting tools with short examples.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Basics

A **figure** is the whole canvas. An **axes** is one plot inside it.

You can plot with `plt.` directly, or create `fig, ax = plt.subplots()` and use `ax.`. The second is better once you have more than one plot.

### Line plot

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating a line plot
plt.plot(x, y)

# Displaying the graph
plt.show()
````

### Same plot, object-oriented, with `figsize`

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating figure with custom dimensions
fig, ax = plt.subplots(figsize=(10, 5))

# Creating data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating line plot
ax.plot(x, y)

# Displaying graph
plt.show()
````

## 2. Line plots

For trends over time.

### Style: `color`, `linestyle`, `marker`

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Creating a red dashed line with circle markers
plt.plot(x, y, color="red", linestyle="--", marker="o")

# Displaying the graph
plt.show()
````

### Title and axis labels

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]
sales = [100, 150, 130, 200, 250, 300]

# Creating the line plot
plt.plot(months, sales, color="blue", marker="o")

# Adding graph title
plt.title("Monthly Sales")

# Naming the X-axis
plt.xlabel("Months")

# Naming the Y-axis
plt.ylabel("Sales")

# Displaying the graph
plt.show()
````

### Two lines

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]

product_a = [100, 150, 130, 200, 250, 300]
product_b = [80, 120, 160, 170, 220, 280]

# Plotting Product A
plt.plot(months, product_a, color="blue", marker="o", label="Product A")

# Plotting Product B
plt.plot(months, product_b, color="red", marker="s", label="Product B")

# Adding title and axis labels
plt.title("Product Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

# Displaying the legend
plt.legend()

# Displaying the graph
plt.show()
````

### Line width, marker size, grid

````python
import matplotlib.pyplot as plt

# Creating the data
days = [1, 2, 3, 4, 5, 6, 7]

temperature = [25, 27, 26, 29, 31, 30, 32]

# Creating the line plot
plt.plot(
    days,
    temperature,
    color="orange",
    linestyle="--",
    marker="o",
    linewidth=2,
    markersize=7,
    label="Temperature"
)

# Adding title and axis labels
plt.title("Weekly Temperature Report")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

# Adding legend and grid
plt.legend()
plt.grid(True, alpha=0.4)

# Displaying the graph
plt.show()
````

### With numpy

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
````

## 4. Histograms

Groups numbers into bins and counts how many fall in each.

### Basic

````python
import matplotlib.pyplot as plt

# Creating sample marks of students
marks = [
    35, 40, 45, 50, 55, 60, 65, 70,
    72, 75, 78, 80, 82, 85, 88, 90,
    92, 95, 45, 55, 60, 65, 70, 75
]

# Creating the histogram
plt.hist(marks, bins=5, color="skyblue", edgecolor="black")

# Adding title and labels
plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

# Displaying the graph
plt.show()
````

### Positive, negative, no relation

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5, 6]

# Positive relationship
positive_y = [10, 20, 30, 40, 50, 60]

# Negative relationship
negative_y = [60, 50, 40, 30, 20, 10]

# No clear linear relationship
random_y = [25, 55, 15, 60, 35, 45]

# Creating three scatter plots
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# Positive correlation
axes[0].scatter(x, positive_y)
axes[0].set_title("Positive Correlation")

# Negative correlation
axes[1].scatter(x, negative_y)
axes[1].set_title("Negative Correlation")

# No clear linear correlation
axes[2].scatter(x, random_y)
axes[2].set_title("No Clear Linear Trend")

# Displaying all three plots
plt.tight_layout()
plt.show()
````

## 7. Subplots

`plt.subplots(rows, cols)` returns the figure and the axes. In a grid, use `axes[row, col]`.

### Side by side

````python
# Importing Matplotlib
import matplotlib.pyplot as plt

# Creating two plotting areas
fig, axes = plt.subplots(1, 2)

# First dataset
x = [1, 2, 3, 4]
y1 = [2, 4, 6, 8]

# Second dataset
y2 = [1, 4, 9, 16]

# Plotting on first axes
axes[0].plot(x, y1)

# Plotting on second axes
axes[1].plot(x, y2)

# Displaying figure
plt.show()
````

### Same data, two years

The y-axes differ. `sharey` fixes this (see below).

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5]

sales_2025 = [100, 150, 200, 180, 250]
sales_2026 = [120, 170, 220, 240, 300]

# Creating one row and two columns of plots
fig, axes = plt.subplots(1, 2)

# Creating the first graph
axes[0].plot(months, sales_2025)
axes[0].set_title("Sales 2025")

# Creating the second graph
axes[1].plot(months, sales_2026)
axes[1].set_title("Sales 2026")

# Displaying both graphs
plt.show()
````

### 2x2 with data

`tight_layout()` stops titles and labels overlapping.

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]
y3 = [10, 25, 15, 35, 30]
y4 = [5, 15, 25, 35, 45]

# Creating a 2x2 subplot layout
fig, axes = plt.subplots(2, 2)

# First graph
axes[0, 0].plot(x, y1)
axes[0, 0].set_title("Increasing Data")

# Second graph
axes[0, 1].plot(x, y2)
axes[0, 1].set_title("Decreasing Data")

# Third graph
axes[1, 0].plot(x, y3)
axes[1, 0].set_title("Variable Data")

# Fourth graph
axes[1, 1].plot(x, y4)
axes[1, 1].set_title("Another Dataset")

# Adjusting the spacing
plt.tight_layout()

# Displaying the figure
plt.show()
````

### Shared axes: `sharex`, `sharey`

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [15, 25, 35, 45, 55]

# Creating two subplots with shared axes
fig, axes = plt.subplots(
    2,
    1,
    sharex=True,
    sharey=True
)

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Dataset A")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Dataset B")

# Adding a shared X-axis label
fig.supxlabel("X Values")

# Adding a shared Y-axis label
fig.supylabel("Y Values")

plt.tight_layout()

plt.show()
````

### Title for the whole figure: `suptitle`

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [50, 40, 30, 20, 10]

# Creating two plots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First graph
axes[0].plot(x, y1)
axes[0].set_title("Increasing")

# Second graph
axes[1].plot(x, y2)
axes[1].set_title("Decreasing")

# Adding the main figure title
fig.suptitle("Data Comparison")

# Adjusting layout to make room for the main title
fig.tight_layout(rect=[0, 0, 1, 0.93])

plt.show()
````

## 8. Annotations

`plt.annotate(text, xy, xytext, arrowprops)`. `xy` is the point, `xytext` is where the text goes.

### Basic

````python
import matplotlib.pyplot as plt

# Creating the data
months = [1, 2, 3, 4, 5, 6]
sales = [100, 150, 130, 200, 250, 300]

# Creating the line plot
plt.plot(months, sales, marker="o")

# Adding annotation
plt.annotate(
    "Highest Sales",
    xy=(6, 300),
    xytext=(4, 320),
    arrowprops={"arrowstyle": "->"}
)

# Adding title and labels
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
````

### Styled

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 40, 30]

plt.figure(figsize=(8, 5))

plt.plot(
    months,
    sales,
    marker="o",
    color="blue"
)

# Highlight the highest point
plt.annotate(
    "Maximum Sales",
    xy=(4, 40),
    xytext=(3, 48),
    fontsize=12,
    fontweight="bold",
    color="red",
    arrowprops=dict(
        arrowstyle="->",
        color="red",
        linewidth=1.5
    )
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
````

### Highlight a point with `scatter`

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [10, 20, 15, 40, 30]

# Main line
plt.plot(months, sales, color="blue", marker="o")

# Highlight maximum value
plt.scatter(
    4,
    40,
    color="red",
    s=150,
    zorder=5
)

plt.title("Sales Performance")

plt.show()
````

## 9. Saving

Call `plt.savefig()` before `plt.show()`. `dpi=300` gives a sharp image. `bbox_inches="tight"` stops labels being cut off.

````python
import matplotlib.pyplot as plt

# Creating the data
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

# Creating the plot
plt.plot(x, y, marker="o")

# Adding title
plt.title("My First Saved Graph")

# Saving the graph
plt.savefig("my_graph.png", dpi=300, bbox_inches="tight")

# Displaying the graph
plt.show()
````

````python
import matplotlib.pyplot as plt

# Creating the data
activities = [
    "Sleep",
    "College",
    "Study",
    "Exercise",
    "Entertainment",
    "Other"
]

hours = [8, 6, 4, 1, 3, 2]

# Creating the chart
plt.figure(figsize=(8, 8))

# Creating the donut chart
plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

# Adding title
plt.title("My Daily Time Distribution")

# Keeping the chart circular
plt.axis("equal")

# Saving the chart
plt.savefig("daily_time.png", dpi=300, bbox_inches="tight")

# Displaying the chart
plt.show()
````

## 11. Full examples

### Styled line chart

````python
import matplotlib.pyplot as plt

# Sample monthly revenue data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [12000, 18000, 15000, 25000, 30000, 40000]

# Apply a built-in style
plt.style.use("seaborn-v0_8-whitegrid")

# Create figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Set background colors
fig.set_facecolor("#F5F5F5")
ax.set_facecolor("#FFFFFF")

# Create line chart
ax.plot(
    months,
    revenue,
    color="#3949AB",
    linewidth=2.5,
    marker="o",
    markersize=7
)

# Format title
ax.set_title(
    "Monthly Revenue Growth",
    fontsize=20,
    fontweight="bold",
    pad=20
)

# Format axis labels
ax.set_xlabel("Months", fontsize=13)
ax.set_ylabel("Revenue (₹)", fontsize=13)

# Set Y-axis limits
ax.set_ylim(0, 45000)

# Format ticks
ax.tick_params(axis="both", labelsize=11)

# Customize grid
ax.grid(
    True,
    linestyle="--",
    linewidth=0.6,
    alpha=0.5
)

# Remove unnecessary spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Adjust layout
plt.tight_layout()

# Display chart
plt.show()
````

### Two lines, legend, annotation

````python
import matplotlib.pyplot as plt

# Sample data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = [100, 150, 130, 200, 250, 300]
product_b = [80, 100, 120, 150, 180, 220]

# Create figure
fig, ax = plt.subplots(figsize=(10, 6))

# Plot Product A
ax.plot(
    months,
    product_a,
    marker="o",
    linewidth=2,
    label="Product A"
)

# Plot Product B
ax.plot(
    months,
    product_b,
    marker="s",
    linewidth=2,
    label="Product B"
)

# Add chart title
ax.set_title(
    "Product Performance Analysis",
    fontsize=18,
    fontweight="bold"
)

# Add axis labels
ax.set_xlabel("Months", fontsize=12)
ax.set_ylabel("Sales", fontsize=12)

# Add legend
ax.legend(
    title="Products",
    loc="upper left",
    frameon=False
)

# Highlight Product A's highest sales
ax.annotate(
    "Peak Sales",
    xy=("Jun", 300),
    xytext=("Apr", 330),
    arrowprops=dict(
        arrowstyle="->",
        color="red"
    ),
    color="red",
    fontweight="bold"
)

# Add grid
ax.grid(
    True,
    linestyle="--",
    alpha=0.4
)

# Adjust layout
plt.tight_layout()

# Display chart
plt.show()
````

### Dashboard

````python
# ==========================================
# PERSONAL DATA ANALYSIS DASHBOARD
# Library: Matplotlib
# ==========================================

import matplotlib.pyplot as plt

# ------------------------------------------
# 1. DATASET
# ------------------------------------------

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

revenue = [12000, 15000, 13500, 18000, 22000, 25000]

expenses = [8000, 9000, 8500, 11000, 13000, 14000]

profit = [4000, 6000, 5000, 7000, 9000, 11000]

customers = [120, 150, 140, 180, 220, 260]


# ------------------------------------------
# 2. CREATE FIGURE AND SUBPLOTS
# ------------------------------------------

fig, ax = plt.subplots(2, 2, figsize=(14, 9))

# Main dashboard title
fig.suptitle(
    "PERSONAL BUSINESS DATA ANALYSIS DASHBOARD",
    fontsize=16,
    fontweight="bold"
)


# ------------------------------------------
# 3. LINE PLOT - REVENUE GROWTH
# ------------------------------------------

ax[0, 0].plot(
    months,
    revenue,
    marker="o",
    color="blue",
    linewidth=2,
    label="Revenue"
)

ax[0, 0].set_title("Monthly Revenue Growth")
ax[0, 0].set_xlabel("Months")
ax[0, 0].set_ylabel("Revenue (₹)")

ax[0, 0].legend()
ax[0, 0].grid(True)


# ------------------------------------------
# 4. BAR CHART - MONTHLY EXPENSES
# ------------------------------------------

ax[0, 1].bar(
    months,
    expenses,
    color="orange",
    label="Expenses"
)

ax[0, 1].set_title("Monthly Expenses")
ax[0, 1].set_xlabel("Months")
ax[0, 1].set_ylabel("Expenses (₹)")

ax[0, 1].legend()
ax[0, 1].grid(axis="y")


# ------------------------------------------
# 5. SCATTER PLOT - CUSTOMERS VS REVENUE
# ------------------------------------------

ax[1, 0].scatter(
    customers,
    revenue,
    color="green",
    marker="o",
    s=80,
    label="Customer Data"
)

ax[1, 0].set_title("Customers vs Revenue")
ax[1, 0].set_xlabel("Number of Customers")
ax[1, 0].set_ylabel("Revenue (₹)")

ax[1, 0].legend()
ax[1, 0].grid(True)


# ------------------------------------------
# 6. PIE CHART - TOTAL REVENUE VS EXPENSES
# ------------------------------------------

total_revenue = sum(revenue)
total_expenses = sum(expenses)

labels = ["Revenue", "Expenses"]

values = [total_revenue, total_expenses]

ax[1, 1].pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.05, 0)
)

ax[1, 1].set_title("Total Revenue vs Expenses")


# ------------------------------------------
# 7. FINAL DASHBOARD CUSTOMIZATION
# ------------------------------------------

# Adjust spacing between subplots
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save dashboard as an image
plt.savefig(
    "personal_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the dashboard
plt.show()


# ------------------------------------------
# 8. BASIC BUSINESS SUMMARY
# ------------------------------------------

print("\n========== BUSINESS SUMMARY ==========")

print(f"Total Revenue: ₹{total_revenue}")
print(f"Total Expenses: ₹{total_expenses}")
print(f"Total Profit: ₹{sum(profit)}")
print(f"Total Customers: {sum(customers)}")

print("======================================")
````
