# DATATYPE

**A data type is a label that tells a computer what kind of information it is dealing with and what it is allowed to do with it.** Think of data types like labeled storage containers in a kitchen: you put liquids in a bottle, spices in a jar, and bread in a box. Just as you wouldn't pour water into a breadbox, a computer uses data types to keep information organized and prevent errors.

```python
x = 1
y = 2.3
z = "str"
b = True # False
print(type(x),type(y),type(z),type(b))
```

## 1. Numeric Types
- `int`: whole numbers (positive, negative, zero), with no size limit in Python
- `float`: numbers with a decimal point
- `complex`: numbers with a real and imaginary part (used in engineering and signal processing)

```python
a = 42          # int
b = -7          # int
c = 3.14        # float
d = 2e3         # float (scientific notation = 2000.0)
e = 3 + 4j      # complex

print(type(a), type(c), type(e))
print(e.real, e.imag)       # 3.0 4.0
print(10 / 3, 10 // 3, 10 % 3, 2 ** 5)   # 3.333..., 3, 1, 32
```

## 2. String (`str`)
Text inside quotes. Strings are **immutable**: you can't change them in place.

```python
name = "Python"
print(name[0], name[-1], name[0:3])    # P n Pyt
print(name.upper(), name.lower(), len(name))
print("Hello " + name, "ha" * 3)       # concatenation, repetition
print(f"I am learning {name}")         # f-string

# name[0] = "J"   # TypeError: strings are immutable
```

## 3. Boolean (`bool`)
Only two values: `True` and `False`. They come from comparisons and are used in `if` and loops.

```python
print(5 > 3, 5 == 3, "a" in "apple")    # True False True
print(True and False, True or False, not True)

# Truthy / Falsy values
print(bool(0), bool(""), bool([]), bool(None))   # all False
print(bool(1), bool("hi"), bool([0]))            # all True
print(True + True)   # 2, because bool is a subclass of int
```

## 4. Collection Types
| Type | Syntax | Ordered | Mutable | Duplicates |
|------|--------|---------|---------|------------|
| `list` | `[1, 2]` | Yes | Yes | Allowed |
| `tuple` | `(1, 2)` | Yes | No | Allowed |
| `set` | `{1, 2}` | No | Yes | Not allowed |
| `dict` | `{"a": 1}` | Yes (3.7+) | Yes | Keys must be unique |

```python
my_list  = [1, 2, 3, 3]
my_tuple = (1, 2, 3, 3)
my_set   = {1, 2, 3, 3}
my_dict  = {"name": "AVU", "age": 21}

my_list.append(4)
print(my_list)       # [1, 2, 3, 3, 4]
print(my_set)        # {1, 2, 3}  (duplicate removed)
print(my_dict["name"])
my_dict["city"] = "Ghaziabad"
print(my_dict)

print(type(my_list), type(my_tuple), type(my_set), type(my_dict))
```

## 5. NoneType
`None` means "no value". It's used as a placeholder for something that doesn't exist yet.

```python
result = None
print(result, type(result))     # None <class 'NoneType'>
print(result is None)           # True (use `is`, not `==`)
```

## 6. Type Conversion (Casting)
You can convert one type into another with `int()`, `float()`, `str()`, `bool()`, `list()`, etc.

```python
print(int("25") + 5)         # 30
print(float(3))              # 3.0
print(int(7.9))              # 7  (truncates, doesn't round)
print(str(100) + " marks")   # 100 marks
print(list("abc"))           # ['a', 'b', 'c']
print(set([1, 1, 2]))        # {1, 2}

# int("hello")   # ValueError: can't convert
```

## 7. Checking Types
Use `type()` to see the exact type, and `isinstance()` to check whether a value belongs to a type.

```python
x = 10
print(type(x))                      # <class 'int'>
print(isinstance(x, int))           # True
print(isinstance(x, (int, float)))  # True (checks multiple types)
print(isinstance(True, int))        # True
```

## 8. Mutable vs Immutable
- **Immutable** (can't change after creation): `int`, `float`, `str`, `bool`, `tuple`
- **Mutable** (can change in place): `list`, `set`, `dict`

```python
a = [1, 2, 3]
b = a            # b points to the SAME list
b.append(99)
print(a)         # [1, 2, 3, 99], so a changed too!

c = a.copy()     # a real copy
c.append(100)
print(a, c)
```
